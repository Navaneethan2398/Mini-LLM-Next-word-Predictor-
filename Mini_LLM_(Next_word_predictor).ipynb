```python
import torch
import torch.nn as nn
import torch.optim as optim
```

Step 1 : Small dataset

```python
sentences = [
    "i love deep learning",
    "deep learning is fun",
    "i love ai",
    "ai is the future"
]
```

Step 2 : Tokenization

```python
words = []

for sentence in sentences:
    words.extend(sentence.split())

vocab = sorted(set(words))

word_to_idx = {word: idx for idx, word in enumerate(vocab)}
idx_to_word = {idx: word for word, idx in word_to_idx.items()}

vocab_size = len(vocab)

print("Vocabulary:", vocab)
print("Vocabulary Size:", vocab_size)
```

Step 3: Create Training Data

```python
X = []
Y = []

for sentence in sentences:
    tokens = sentence.split()

    for i in range(len(tokens) - 1):
        input_word = tokens[i]
        target_word = tokens[i + 1]

        X.append(word_to_idx[input_word])
        Y.append(word_to_idx[target_word])

X = torch.tensor(X)
Y = torch.tensor(Y)

print("\nTraining Inputs:", X)
print("Training Targets:", Y)
```

Step : Build Model

```python
class NextWordPredictor(nn.Module):

    def __init__(self, vocab_size, embedding_dim):
        super().__init__()

        # Embedding layer
        self.embedding = nn.Embedding(vocab_size, embedding_dim)

        # Linear layer
        self.fc = nn.Linear(embedding_dim, vocab_size)

    def forward(self, x):

        # Convert word IDs into embeddings
        embedded = self.embedding(x)

        # Predict next word scores
        output = self.fc(embedded)

        return output
```

Step 5: initialize Model

```python
embedding_dim = 10

model = NextWordPredictor(vocab_size, embedding_dim)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

Step 6 : Train Model

```python
epochs = 500

for epoch in range(epochs):

    optimizer.zero_grad()

    outputs = model(X)

    loss = criterion(outputs, Y)

    loss.backward()

    optimizer.step()

    if (epoch + 1) % 50 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```

Step 7 : Predict Next Word

```python

def predict_next_word(word):

    model.eval()

    input_idx = torch.tensor([word_to_idx[word]])

    with torch.no_grad():

        output = model(input_idx)

        predicted_idx = torch.argmax(output, dim=1).item()

    return idx_to_word[predicted_idx]
```

Step 8 : Test predictions

```python
test_words = ["the", "deep", "ai"]

print("\nNext Word Predictions:")

for word in test_words:
    predicted = predict_next_word(word)
    print(f"{word} --> {predicted}")
```
